```python
from google.colab import files

uploaded = files.upload()

```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv(list(uploaded.keys())[0])


# 1. Bar Chart - Course Category by Platform

sns.countplot(
    x="Category",
    hue="Platform",
    data=df
)

plt.title("Online Courses by Category and Platform")
plt.xlabel("Course Category")
plt.ylabel("Number of Courses")
plt.xticks(rotation=45)

plt.show()


# 2. Pie Chart - Course Category Distribution

category_count = df["Category"].value_counts()

plt.pie(
    category_count,
    labels=category_count.index,
    autopct="%1.1f%%"
)

plt.title("Course Category Distribution")

plt.show()


# 3. Line Chart - Duration vs Enrolled Students

df.groupby("Duration (hours)")["Enrolled_Students"].mean().plot(
    kind="line",
    marker="o"
)

plt.title("Average Enrollment by Course Duration")
plt.xlabel("Course Duration (Hours)")
plt.ylabel("Average Enrolled Students")
plt.grid()

plt.show()
```
