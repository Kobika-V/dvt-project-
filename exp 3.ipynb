```python
import pandas as pd

# 1. Load dataset
df = pd.read_csv(list(uploaded.keys())[0])


# 2. View DataFrame

print("First 5 rows:")
print(df.head())

print("\nLast 5 rows:")
print(df.tail())


# 3. Inspect DataFrame

print("\nShape:")
print(df.shape)

print("\nColumns:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)


# 4. Filtering data

print("\nCourses with Duration > 20 hours:")
print(
    df[df["Duration (hours)"] > 20].head()
)

print("\nCourses with Enrolled Students > 5000:")
print(
    df[df["Enrolled_Students"] > 5000].head()
)

print("\nCourses with Rating > 4:")
print(
    df[df["Rating (out of 5)"] > 4].head()
)


# 5. Select specific columns

print("\nCourse Name and Price:")
print(
    df[["Course_Name", "Price ($)"]].head()
)


# 6. Mean

mean_value = df["Price ($)"].mean()

print("\nMean Price:", mean_value)


# 7. Median

median_value = df["Price ($)"].median()

print("Median Price:", median_value)


# 8. Mode

mode_value = df["Price ($)"].mode()[0]

print("Mode Price:", mode_value)


# 9. Range

data_range = (
    df["Price ($)"].max()
    - df["Price ($)"].min()
)

print("Range of Price:", data_range)


# 10. Variance

variance = df["Price ($)"].var()

print("Variance of Price:", variance)


# 11. Standard Deviation

std = df["Price ($)"].std()

print("Standard Deviation of Price:", std)


# 12. Complete descriptive statistics

print("\nDescriptive Statistics:")

print(
    df["Price ($)"].describe()
)
```
