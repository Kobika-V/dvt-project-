```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler


# 1. Load dataset
df = pd.read_csv(list(uploaded.keys())[0])

print("Original Data:")
print(df.head())


# 2. Check missing values
print("\nMissing Values:")
print(df.isnull().sum())


# 3. Convert numerical columns to numeric

df["Duration (hours)"] = pd.to_numeric(
    df["Duration (hours)"], errors="coerce"
)

df["Enrolled_Students"] = pd.to_numeric(
    df["Enrolled_Students"], errors="coerce"
)

df["Completion_Rate (%)"] = pd.to_numeric(
    df["Completion_Rate (%)"], errors="coerce"
)

df["Price ($)"] = pd.to_numeric(
    df["Price ($)"], errors="coerce"
)

df["Rating (out of 5)"] = pd.to_numeric(
    df["Rating (out of 5)"], errors="coerce"
)


# 4. Fill missing values

df["Duration (hours)"] = df["Duration (hours)"].fillna(
    df["Duration (hours)"].median()
)

df["Enrolled_Students"] = df["Enrolled_Students"].fillna(
    df["Enrolled_Students"].median()
)

df["Completion_Rate (%)"] = df["Completion_Rate (%)"].fillna(
    df["Completion_Rate (%)"].median()
)

df["Price ($)"] = df["Price ($)"].fillna(
    df["Price ($)"].median()
)

df["Rating (out of 5)"] = df["Rating (out of 5)"].fillna(
    df["Rating (out of 5)"].median()
)


# 5. Check duplicates
print("\nDuplicate Rows:", df.duplicated().sum())


# 6. Remove duplicates
df = df.drop_duplicates()


# 7. Remove unnecessary columns

df = df.drop(
    ["Course_ID", "Course_Name"],
    axis=1
)


# 8. Check data types
print("\nData Types:")
print(df.dtypes)


# 9. Min-Max Scaling

scaler = MinMaxScaler()

numeric_columns = [
    "Duration (hours)",
    "Enrolled_Students",
    "Completion_Rate (%)",
    "Price ($)",
    "Rating (out of 5)"
]

df[numeric_columns] = scaler.fit_transform(
    df[numeric_columns]
)


print("\nAfter Min-Max Scaling:")
print(df[numeric_columns].head())


# 10. Standardization

standard_scaler = StandardScaler()

df[numeric_columns] = standard_scaler.fit_transform(
    df[numeric_columns]
)


print("\nAfter Standardization:")
print(df[numeric_columns].head())


# 11. Final dataset information

print("\nFinal Dataset:")
df.info()


print("\nData Cleaning Completed!")
```
