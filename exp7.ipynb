```python
from google.colab import files

uploaded = files.upload()

```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv(list(uploaded.keys())[0])


# 1. Create calculated field

df["Course Status"] = df["Rating (out of 5)"].apply(
    lambda x: "Highly Rated" if x >= 4 else "Lower Rated"
)

print(df[["Rating (out of 5)", "Course Status"]].head())


# 2. Bar Chart - Course Status by Category

sns.countplot(
    x="Category",
    hue="Course Status",
    data=df
)

plt.title("Course Rating Status by Category")
plt.xlabel("Course Category")
plt.ylabel("Number of Courses")
plt.xticks(rotation=45)

plt.show()


# 3. Line Chart - Average Enrollment by Duration

enrollment = df.groupby(
    "Duration (hours)"
)["Enrolled_Students"].mean()

plt.plot(
    enrollment.index,
    enrollment.values,
    marker="o"
)

plt.title("Average Enrollment by Course Duration")
plt.xlabel("Course Duration (Hours)")
plt.ylabel("Average Enrolled Students")

plt.grid()

plt.show()


# 4. Pie Chart - Course Status

status = df["Course Status"].value_counts()

plt.pie(
    status,
    labels=status.index,
    autopct="%1.1f%%"
)

plt.title("Course Rating Status Distribution")

plt.show()
```
